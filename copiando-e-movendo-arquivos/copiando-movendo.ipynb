```python
from pathlib import Path 
import shutil 
import os
```

### Copiando arquivos de formas diferentes

```python
# Copiando arquivo com copyfile 
# Info importante sobre copyfile: O arquivo é copiado, mas sem replicar permissões
pasta_atual = Path.cwd() 

caminho_arquivo = pasta_atual / 'texto.txt' 

destino_copia = pasta_atual / 'destino 1' / 'texto_copiado.txt'

shutil.copyfile(caminho_arquivo, destino_copia)
# Copiando com copy 
# shutil.copy replica permissões também
destino_copia = pasta_atual / 'destino 2' 

shutil.copy(caminho_arquivo, destino_copia)

# Copiando com copy 2 
# copy2 também copia metadados 
destino_copia = pasta_atual / 'destino 3' 

shutil.copy(caminho_arquivo, destino_copia)
```

### Movendo e deletando

```python
# Movendo arquivos 
arquivo_move = pasta_atual / 'move.txt' 
destino_move = pasta_atual / 'destino 1' 

shutil.move(arquivo_move, destino_move)
```

```python
# Deletando arquivos
caminho_move = destino_move / 'move.txt'

if caminho_move.exists:
    os.remove(caminho_move)
```
