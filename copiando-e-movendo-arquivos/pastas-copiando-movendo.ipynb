```python
from pathlib import Path 
import shutil
```

# Criando pasta 

```python
pasta_atual = Path.cwd()

caminho_pasta_destino = pasta_atual / "destino 4"

# flag exists_ok=True evita erro se a pasta já tiver sido criada, e mantém a criação caso ainda não exista
caminho_pasta_destino.mkdir(exist_ok=True) 
```

# Criando pasta com subpasta

```python
pasta_atual = Path.cwd()

caminho_pasta_destino = pasta_atual / "destino 5" / "destino 5b"

# flag parents=True permite a criação de pasta com subpastas dentro.
caminho_pasta_destino.mkdir(parents=True) 
```

# Copiando pastas

```python
pasta_atual = Path.cwd()

pasta_copiada = pasta_atual / "destino 5"

destino_copia = pasta_atual / "destino 1" / "destino 5 copy"

shutil.copytree(pasta_copiada, destino_copia)
```
